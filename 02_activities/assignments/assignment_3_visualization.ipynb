## Visualization 1: Python

### Load dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# load data from csv file
website_visits_df = pd.read_csv('../../04_this_cohort/data/Visits by Day of Week.csv')
website_visits_df
```

### Clean Dataset

```python
cleaned_visits = website_visits_df.dropna() # drop first row with NA day of week

# Reorder days of week from Monday to Sunday
day_of_week_ordered = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'] 
cleaned_visits = cleaned_visits.groupby(['Day of Week']).sum().reindex(day_of_week_ordered)

cleaned_visits
```

### Create Visualization

```python
# set x and y variables
x = cleaned_visits.index
y = cleaned_visits['Sessions']

# create font libraries for tile, x label and y label
font1 = {'family':'arial', 'weight':'bold', 'size':20}
font2 = {'family':'arial', 'size':16}

# create figure
fit, ax = plt.subplots(figsize=(10,5), 
                       layout='constrained')

## create bar graph
ax.bar(x,y, color=('royalblue', 'royalblue','royalblue','royalblue','royalblue','orange','orange')) ## differentiate colour of week days and weekend days
ax.set_title('City of Toronto Daily Website Traffic (2024)', fontdict=font1)
ax.set_ylabel('Sessions', fontdict=font2)
ax.set_xlabel('Day of Week', fontdict=font2)
ax.tick_params(labelsize=12) # increase font size of ticks to 12

# add lengend to differentiate colours of week days vs weekend days
legend_elements = [Patch(facecolor='royalblue', label='Weekday'),
                   Patch(facecolor='orange', label='Weekend')]

ax.legend(handles=legend_elements, loc='upper right', fontsize=12)
```
