```python
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(['Luffy','Zoro','Nami','Usopp','Sanji'])
y2 = np.array([110,180,240,99,220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize=(10,3), 
                               layout = 'constrained')

font1 = {'family':'serif', 'size':14, 'color': "#411A54"}
font2 = {'family':'arial', 'size':10, 'color': "#950835"}

ax1.scatter(x1,y1,
            marker = 'x', 
            color = 'red')
ax1.set_title('Title 1', fontdict=font1)
ax1.set_ylabel('y Axis',fontdict=font2) 
ax1.set_xlabel('x Axis',fontdict=font2)

ax2.bar(x2,y2, 
        color = 'pink')
ax2.set_title('Title 2', fontdict=font1)
ax2.set_ylabel('y Axis',fontdict=font2) 
ax2.set_xlabel('x Axis',fontdict=font2)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], 
                                    ['ax2', 'ax3']], 
                                    figsize=(7, 4),
                                    layout = 'constrained')

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
```
